```python
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
# Step 1: Create Bayesian Network structure
model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm"),
    ("Alarm", "JohnCalls"),
    ("Alarm", "MaryCalls")
])


# Step 2: Define Conditional Probability Tables (CPDs)

# P(Burglary)
cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.99], [0.01]],
    state_names={"Burglary": ["False", "True"]}
)

# P(Earthquake)
cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)

# P(Alarm | Burglary, Earthquake)
cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],
        [0.001, 0.29, 0.94, 0.95]
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)

# P(JohnCalls | Alarm)
cpd_john = TabularCPD(
    variable="JohnCalls",
    variable_card=2,
    values=[
        [0.95, 0.10],
        [0.05, 0.90]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "JohnCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)

# P(MaryCalls | Alarm)
cpd_mary = TabularCPD(
    variable="MaryCalls",
    variable_card=2,
    values=[
        [0.99, 0.30],
        [0.01, 0.70]
    ],
    evidence=["Alarm"],
    evidence_card=[2],
    state_names={
        "MaryCalls": ["False", "True"],
        "Alarm": ["False", "True"]
    }
)


# Step 3: Add CPDs to model
model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm,
    cpd_john,
    cpd_mary
)


# Step 4: Check model correctness
print("Is Bayesian Network valid?", model.check_model())


# Step 5: Perform inference
inference = VariableElimination(model)


# Query: Probability of Burglary when John and Mary both call
result = inference.query(
    variables=["Burglary"],
    evidence={
        "JohnCalls": "True",
        "MaryCalls": "True"
    }
)

print("\nProbability of Burglary given JohnCalls=True and MaryCalls=True:")
print(result)
print("Afsha 25")



```

```python
# Posterior Probability using Bayesian Network
# Query: P(Burglary | Alarm = True)

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination


# Step 1: Create Bayesian Network structure

model = DiscreteBayesianNetwork([
    ("Burglary", "Alarm"),
    ("Earthquake", "Alarm")
])


# Step 2: Define probability of Burglary

cpd_burglary = TabularCPD(
    variable="Burglary",
    variable_card=2,
    values=[[0.999], [0.001]],
    state_names={"Burglary": ["False", "True"]}
)


# Step 3: Define probability of Earthquake

cpd_earthquake = TabularCPD(
    variable="Earthquake",
    variable_card=2,
    values=[[0.998], [0.002]],
    state_names={"Earthquake": ["False", "True"]}
)


# Step 4: Define probability of Alarm
# P(Alarm | Burglary, Earthquake)

cpd_alarm = TabularCPD(
    variable="Alarm",
    variable_card=2,
    values=[
        [0.999, 0.71, 0.06, 0.05],  # Alarm = False
        [0.001, 0.29, 0.94, 0.95]   # Alarm = True
    ],
    evidence=["Burglary", "Earthquake"],
    evidence_card=[2, 2],
    state_names={
        "Alarm": ["False", "True"],
        "Burglary": ["False", "True"],
        "Earthquake": ["False", "True"]
    }
)


# Step 5: Add CPDs to model

model.add_cpds(
    cpd_burglary,
    cpd_earthquake,
    cpd_alarm
)


# Step 6: Check whether model is correct

print("Is Bayesian Network valid?", model.check_model())


# Step 7: Perform inference

inference = VariableElimination(model)

result = inference.query(
    variables=["Burglary"],
    evidence={"Alarm": "True"}
)


print("\nPosterior Probability:")
print("P(Burglary | Alarm = True)")
print(result)
print("Afsha 25")
```

```python
# Posterior Probability using Bayesian Network
# Query: P(Rain | WetGrass = True)

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination


# Step 1: Create Bayesian Network structure

model = DiscreteBayesianNetwork([
    ("Rain", "WetGrass"),
    ("Sprinkler", "WetGrass")
])


# Step 2: Define probability of Rain

cpd_rain = TabularCPD(
    variable="Rain",
    variable_card=2,
    values=[[0.7],
            [0.3]],
    state_names={
        "Rain": ["False", "True"]
    }
)


# Step 3: Define probability of Sprinkler

cpd_sprinkler = TabularCPD(
    variable="Sprinkler",
    variable_card=2,
    values=[[0.6],
            [0.4]],
    state_names={
        "Sprinkler": ["False", "True"]
    }
)


# Step 4: Define probability of Wet Grass
# P(WetGrass | Rain, Sprinkler)

cpd_wetgrass = TabularCPD(
    variable="WetGrass",
    variable_card=2,
    values=[
        [1.00, 0.20, 0.10, 0.01],   # WetGrass = False
        [0.00, 0.80, 0.90, 0.99]    # WetGrass = True
    ],
    evidence=["Rain", "Sprinkler"],
    evidence_card=[2, 2],
    state_names={
        "WetGrass": ["False", "True"],
        "Rain": ["False", "True"],
        "Sprinkler": ["False", "True"]
    }
)


# Step 5: Add CPDs to the model

model.add_cpds(
    cpd_rain,
    cpd_sprinkler,
    cpd_wetgrass
)


# Step 6: Check whether the model is correct

print("Is Bayesian Network valid?", model.check_model())


# Step 7: Perform inference

inference = VariableElimination(model)


# Step 8: Compute posterior probability
# P(Rain | WetGrass = True)

result = inference.query(
    variables=["Rain"],
    evidence={
        "WetGrass": "True"
    }
)


# Step 9: Display the result

print("\nPosterior Probability:")
print("P(Rain | WetGrass = True)")
print(result)
print("Afsha 25")
```
